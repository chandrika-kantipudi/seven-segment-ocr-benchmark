```python

import os
import pandas as pd
from sklearn.model_selection import train_test_split

dataset_dir = "/content/seven-segment-ocr/data/ocr_training_data"
gt_path = os.path.join(dataset_dir, "gt.txt")

# Read image filenames and their correct text labels
df = pd.read_csv(
    gt_path,
    sep="\t",
    header=None,
    names=["image", "text"],
    dtype=str
)

# Keep only records whose image files exist
df["image_path"] = df["image"].apply(
    lambda name: os.path.join(dataset_dir, name)
)
df = df[df["image_path"].apply(os.path.isfile)].reset_index(drop=True)

# 70% train, 30% temporary
train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    shuffle=True
)

# Split the remaining 30% equally into validation and test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    shuffle=True
)

print("Total images:", len(df))
print("Training images:", len(train_df))
print("Validation images:", len(val_df))
print("Test images:", len(test_df))

print("\nSample test records:")
print(test_df[["image", "text"]].head())
```

```python

import pandas as pd
import numpy as np
import re

from jiwer import cer, wer

results_path = "/content/ppocr_test_results.csv"
results_df = pd.read_csv(
    results_path,
    dtype={"ground_truth": str, "prediction": str}
).fillna("")

def normalize_text(text):
    # Remove whitespace and common unit text, preserving digits and decimals
    text = str(text).strip().replace(" ", "")
    text = re.sub(r"(?i)kwh", "", text)
    return text

y_true = results_df["ground_truth"].map(normalize_text).tolist()
y_pred = results_df["prediction"].map(normalize_text).tolist()

# Accuracy and error metrics
exact_matches = sum(a == b for a, b in zip(y_true, y_pred))
exact_match_accuracy = exact_matches / len(y_true) * 100
cer_value = cer(y_true, y_pred) * 100
wer_value = wer(y_true, y_pred) * 100

# Latency metrics (milliseconds)
latencies = results_df["latency_ms"].astype(float)

print("===== PP-OCR TEST RESULTS =====")
print("Test images:", len(results_df))
print(f"Exact matches: {exact_matches}/{len(results_df)}")
print(f"Exact-match accuracy: {exact_match_accuracy:.2f}%")
print(f"CER: {cer_value:.2f}%")
print(f"WER: {wer_value:.2f}%")
print(f"Mean latency: {latencies.mean():.2f} ms")
print(f"P50 latency: {latencies.quantile(0.50):.2f} ms")
print(f"P99 latency: {latencies.quantile(0.99):.2f} ms")

print("\n===== INCORRECT PREDICTIONS =====")
errors = results_df[
    results_df.apply(
        lambda row: normalize_text(row["ground_truth"])
        != normalize_text(row["prediction"]),
        axis=1
    )
]
print(errors[["image", "ground_truth", "prediction"]].to_string(index=False))

results_df.to_csv("/content/ppocr_test_results.csv", index=False)
```

```python

import pandas as pd
import re
from jiwer import cer, wer

# Load both experiment results
baseline = pd.read_csv("/content/ppocr_test_results.csv")
resized = pd.read_csv("/content/ppocr_resized_test_results.csv")

# Use identical text normalization for a fair comparison
def normalize_text(text):
    text = str(text).strip().replace(" ", "")
    text = re.sub(r"(?i)kwh", "", text)
    return text

def calculate_metrics(df, experiment_name):
    actual = df["ground_truth"].map(normalize_text).tolist()
    predicted = df["prediction"].fillna("").map(normalize_text).tolist()

    exact_matches = sum(a == p for a, p in zip(actual, predicted))

    return {
        "Experiment": experiment_name,
        "Exact Matches": f"{exact_matches}/{len(df)}",
        "Exact Accuracy (%)": round(100 * exact_matches / len(df), 2),
        "CER (%)": round(cer(actual, predicted) * 100, 2),
        "WER (%)": round(wer(actual, predicted) * 100, 2),
        "Mean Latency (ms)": round(df["latency_ms"].mean(), 2),
        "P50 Latency (ms)": round(df["latency_ms"].quantile(0.50), 2),
        "P99 Latency (ms)": round(df["latency_ms"].quantile(0.99), 2),
        "Empty Predictions": int(
            df["prediction"].fillna("").str.strip().eq("").sum()
        )
    }

comparison = pd.DataFrame([
    calculate_metrics(baseline, "Original 200x31"),
    calculate_metrics(resized, "Resized 800x124")
])

print("PP-OCR BASELINE vs RESIZED IMAGE COMPARISON")
display(comparison)
```

```python

import pandas as pd
import re
from jiwer import cer, wer

baseline = pd.read_csv(
    "/content/ppocr_test_results.csv",
    dtype=str,
    keep_default_na=False
)

resized = pd.read_csv(
    "/content/ppocr_resized_test_results.csv",
    dtype=str,
    keep_default_na=False
)

def normalize_text(text):
    text = str(text).strip().replace(" ", "")
    text = re.sub(r"(?i)kwh", "", text)
    return text

def calculate_metrics(df, name):
    actual = df["ground_truth"].map(normalize_text).tolist()
    predicted = df["prediction"].map(normalize_text).tolist()

    matches = sum(a == p for a, p in zip(actual, predicted))

    latencies = pd.to_numeric(df["latency_ms"], errors="coerce")

    return {
        "Experiment": name,
        "Exact Matches": f"{matches}/{len(df)}",
        "Exact Accuracy (%)": round(matches / len(df) * 100, 2),
        "CER (%)": round(cer(actual, predicted) * 100, 2),
        "WER (%)": round(wer(actual, predicted) * 100, 2),
        "Mean Latency (ms)": round(latencies.mean(), 2),
        "P50 Latency (ms)": round(latencies.quantile(0.50), 2),
        "P99 Latency (ms)": round(latencies.quantile(0.99), 2),
        "Empty Predictions": sum(p.strip() == "" for p in predicted)
    }

display(pd.DataFrame([
    calculate_metrics(baseline, "Original 200x31"),
    calculate_metrics(resized, "Resized 800x124")
]))

print("\nGround-truth formatting check:")
print(baseline[["ground_truth", "prediction"]].head(10).to_string(index=False))
```
